```python
#Imprimir las 50 primeras líneas
rows=[]
for row in csvreader:
    rows.append(row)

print(rows[:50])
```

```python
from statistics import mean, median, mode, stdev
```

```python
import numpy as np
```

```python
height=np.array(rows)
```

```python
#Generacion de listado de altura de hombres en cm (col 2) y transformación en número
height_h=height[1:,2]
height_h
```

```python
height_h_cm=[float(numeric_string)for numeric_string in height_h]
height_h_cm[0:10]
```

```python
#Generacion de listado de altura de hombres en Ft (col 4) y transformación en número
height_h1=height[1:,4]
height_h1
```

```python
height_h_ft=[float(numeric_string)for numeric_string in height_h1]
height_h_ft[0:10]
```

```python
#Calculo de media, max y min en los datos de los hombres, tanto en cm como en ft
media_h_cm=mean(height_h_cm)
max_h_cm=max(height_h_cm)
min_h_cm=min(height_h_cm)

media_h_ft=mean(height_h_ft)
max_h_ft=max(height_h_ft)
min_h_ft=min(height_h_ft)

print('En promedio, la estatura de los hombres en cm es ',media_h_cm)
print('El valor máximo en la estatura de los hombres en cm es ', max_h_cm)
print('El valor minimo en la estatura de los hombres en cm es ', min_h_cm)

print('En promedio, la estatura de los hombres en ft es ',media_h_ft)
print('El valor máximo en la estatura de los hombres en ft es ', max_h_ft)
print('El valor minimo en la estatura de los hombres en ft es ', min_h_ft)
```
