```python
#Uso de liberia BeautifulSoup

from bs4 import BeautifulSoup
soup = BeautifulSoup(response.text, "html.parser")

# Extraemos todos los enlaces
enlaces = soup.find_all("a")

print(f"Se encontraron {len(enlaces)} enlaces:\n")

# Mostramos los primeros 5
for enlace in enlaces[:5]:
    print(enlace.get("href"))
```

```python
#Generación de un listado
import requests

url = 'http://makeup-api.herokuapp.com/api/v1/products.json'

response = requests.get(url)
productos = response.json()  

lista_productos = [producto["name"] for producto in productos if producto["name"]]

# Mostramos las primeras 10 de forma ordenada
for i, producto in enumerate(lista_productos[:10], 1):
    print(f"{i}. {producto}")
```

```python
import requests

api_url = "http://makeup-api.herokuapp.com/api/v1/products.json"
response = requests.get(api_url)

print("Código de estado:", response.status_code)

productos = response.json()
print(f"\nCantidad total de productos: {len(productos)}")

# Mostramos los primeros 10 productos
print("\nEjemplo de productos:\n")
for i, p in enumerate(productos[:10], 1):
    print(f"{i}. {p['brand']} - {p['name']} | Tipo: {p['product_type']} | Precio: {p['price']}")
```

```python
# Filtrar productos de la marca "maybelline"
marca = "maybelline"
url_marca = f"http://makeup-api.herokuapp.com/api/v1/products.json?brand={marca}"

response = requests.get(url_marca)
productos_marca = response.json()

print(f"\nProductos de la marca '{marca}': {len(productos_marca)}\n")

for i, p in enumerate(productos_marca[:10], 1):
    print(f"{i}. {p['name']} | Tipo: {p['product_type']} | Precio: {p['price']}")
```
