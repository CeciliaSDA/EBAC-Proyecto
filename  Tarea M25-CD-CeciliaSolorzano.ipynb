```python
#Ejercicio raiz cuadrada
def raiz_cuadrada(x):
    return x**0.5
print(raiz_cuadrada(16))
```

```python
#Ejercicio con LAMBDA
raiz_lam=lambda x:x**0.5
print (raiz_lam(16))
```

```python
#Ejercicio determinar largo de una cadena de palabras
cadena = "Hola curso de Analista de Datos"
palabras = cadena.split()
long = []
for palabra in palabras:
    long.append(len(palabra))
print(long)
```

```python
#Ejericio MAP
cadena = "Hola curso de Analista de Datos"
palabras = cadena.split()
long_map=list(map(len,palabras))
print (long_map)
```

```python
#Ejercicio calcular el producto de una lista
from functools import reduce
lista_num = [5, 6, 7, 8]
producto = 1
for num in lista_num:
    producto *= num
print('Resultado:',producto)
```

```python
#Ejercicio con REDUCE
lista_num = [5, 6, 7, 8]
product=reduce(lambda x,y: x*y,lista_num)
print('Resultado:',product)
```

```python
#Ejercicio encontrar palabras qe contengan mayúsculas o numeros en un listado 
listado = ["Cecilia", "palabra", "Solorzano", "apellido", "módulo25", "EBAC",'final']

def contiene_mayus_o_num(palabra):
    for c in palabra:
        if c.isupper() or c.isdigit():
            return True
    return False

palabra_filtradas = []
for palabra in listado:
    if contiene_mayus_o_num(palabra):
        palabra_filtradas.append(palabra)
print(palabra_filtradas)
```

```python
# Ejercicio de FILTER
listado = ["Cecilia", "palabra", "Solorzano", "apellido", "módulo25", "EBAC",'final']
filtradas_filter = list(filter(lambda x: any(c.isupper() or c.isdigit() for c in x), listado))
print(filtradas_filter)
```
