```python
#Exportación de bibliotecas
import pandas as pd
import numpy as np
```

```python
#Importacion de archivos
df_category = pd.read_csv("DIM_CATEGORY.csv")
df_product = pd.read_csv("DIM_PRODUCT.csv")
df_segment = pd.read_csv("DIM_SEGMENT.csv")
df_calendar = pd.read_csv("DIM_CALENDAR.csv")
df_sales = pd.read_csv("FACT_SALES.csv")
```

```python
#Eliminar duplicados
df_product.drop_duplicates(inplace=True)
df_category.drop_duplicates(inplace=True)
df_segment.drop_duplicates(inplace=True)
df_calendar.drop_duplicates(inplace=True)
df_sales.drop_duplicates(inplace=True)
```

```python
#Eliminar nulos
print(df_product.isnull().sum())
df_product.dropna(inplace=True)

print(df_category.isnull().sum())
df_category.dropna(inplace=True)

print(df_segment.isnull().sum())
df_segment.dropna(inplace=True)

print(df_calendar.isnull().sum())
df_calendar.dropna(inplace=True)

print(df_sales.isnull().sum())
df_sales.dropna(inplace=True)
```

```python
#Estandarización de formato de fechas (df_calendar, df_sales)
print(df_calendar["DATE"].head(15))
```

```python
df_calendar["DATE"] = pd.to_datetime(df_calendar["DATE"], format="%Y-%m-%d")
```

```python
#Unir DataFrame (df_product y df_category)
#Limpieza de formatos para evitar errores
df_product["CATEGORY"] = df_product["CATEGORY"].astype(str)
df_category["CATEGORY"] = df_category["CATEGORY"].astype(str)
df_product_cat = pd.merge(df_product, df_category, on="CATEGORY", how="left")
```

```python
#Unir DataFrame (df_product_cat y df_segment)
#limpieza de formatos para evitar errores
cols_to_clean = ["CATEGORY", "FORMAT", "ATTR1", "ATTR2", "ATTR3"]
for col in cols_to_clean:
    df_product_cat[col] = df_product_cat[col].astype(str).str.strip().str.upper()
    df_segment[col] = df_segment[col].astype(str).str.strip().str.upper()
df_product_full = pd.merge(df_product_cat, df_segment, on=["CATEGORY", "FORMAT", "ATTR1", "ATTR2", "ATTR3"], how="left")
```

```python
#Unir DataFrame (df_product_full y df_sale)
#Limpieza de formatos para evitar errores
df_sales["ITEM_CODE"] = df_sales["ITEM_CODE"].astype(str).str.strip().str.upper()
df_product_full["ITEM"] = df_product_full["ITEM"].astype(str).str.strip().str.upper()
df_sales_prod = pd.merge(df_sales, df_product_full,left_on="ITEM_CODE",right_on="ITEM", how="left")
```

```python
#Unir DataFrame (df_sales_product y df_calender)
# Convertir WEEK a string para prevenir errores de tipo 
df_sales_prod["WEEK"] = df_sales_prod["WEEK"].astype(str).str.strip()
df_calendar["WEEK"] = df_calendar["WEEK"].astype(str).str.strip()
df_final = pd.merge(df_sales_prod,df_calendar,on="WEEK",how="left")
```
