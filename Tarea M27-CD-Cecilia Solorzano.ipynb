```python
#Generar Excepciones con TRY Y EXCEPT:
try:
    x = input('Ingresar contraseña: ')
    assert 'Ceci' in x, 'Intenta nuevamente'
    
    if x == 'Cecilia':
        print('Contraseña correcta')
    else:
        print('Cerca, nombre completo')
        
except AssertionError as e:
    print(f'Error: {e}')
```

```python
def numerologia():
    try:
        fecha=input('Ingresa fecha de nacimiento(DD/MM/AAAA): ')
        formato=fecha.split('/')
        assert len(formato)==3, 'El formato debe ser DD/MM/AAAA'

        dia, mes, año= formato
        assert dia.isdigit() and mes.isdigit() and año.isdigit(), 'la fecha solo debe contener numero'
        assert 1<=int(dia) <31,'El dia debe estar entre 1 y 31'
        assert 1<=int(dia) <12, 'El mes debe estar entre 1 y 12'
        assert len(año)==4, 'El año debe tener 4 digitos'

        suma=(sum(int(d) for d in dia+mes+año))
        while suma > 9:
            suma=sum(int(d) for d in str(suma))
        print(f'Tu numero es: {suma}')
        
    except AssertionError as e:
        print(f'Error: {e}')
    except Exception as e:
        print(f'Error inesperado: {e}')
numerologia()
```

```python
def numerologia():
    try:
        fecha=input('Ingresa fecha de nacimiento(DD/MM/AAAA): ')
        formato=fecha.split('/')
        assert len(formato)==3, 'El formato debe ser DD/MM/AAAA'

        dia, mes, año= formato
        assert dia.isdigit() and mes.isdigit() and año.isdigit(), 'la fecha solo debe contener numero'
        assert 1<=int(dia) <31,'El dia debe estar entre 1 y 31'
        assert 1<=int(dia) <12, 'El mes debe estar entre 1 y 12'
        assert len(año)==4, 'El año debe tener 4 digitos'

        suma=(sum(int(d) for d in dia+mes+año))
        while suma > 9:
            suma=sum(int(d) for d in str(suma))
        print(f'Tu numero es: {suma}')
        
    except AssertionError as e:
        print(f'Error: {e}')
    except Exception as e:
        print(f'Error inesperado: {e}')
    else:
        print(f"Tu número de la suerte: {suma}")
    
    finally:
        # Este bloque siempre se ejecuta
        print('Suerte!!')

numerologia()
```
