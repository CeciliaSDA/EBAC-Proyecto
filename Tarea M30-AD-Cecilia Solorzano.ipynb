```python
#Mostrar las primeras 20 filas de archivo, las ultimas 5 y un sample de 10
import pandas as pd
import numpy as np
df=pd.read_csv('fifa_eda.csv')

#20 primeras filas
df[0:20]
```

```python
#Las últimas 5
df.tail(5)
```

```python
#10 filas aleatorias
df.sample(10)
```

```python
#Estadística con .describe() y los tipos de datos de dataset
df.describe()
```

```python
df.dtypes
```

```python
#Añadir una columna "Years Playing", que calcule el año actual menos la columna "Joined"
df['Years Playing']=int(2025)-df['Joined']
```

```python
df.sample(10)
```

```python
#Buscar y mostrar a todos los jugadores de México
df[['ID','Name','Club','Joined']].loc[df['Nationality']=='Mexico']
```

```python
#Ordenar y mostrar los datos por la columna ReleaseClause (sueldo)
df[['ID', 'Name', 'Club', 'Joined', 'Release Clause']].sort_values(by='Release Clause')
```

```python
##Ordenar y mostrar los datos por la columna ReleaseClause (sueldo) de manera descendente
df[['ID', 'Name', 'Club', 'Joined', 'Release Clause']].sort_values(by='Release Clause', ascending=False)
```

```python
#Generar un nuevo dataset que contenga el año (joined) y el número de jugadores (groupby)
df_joined = df.groupby('Joined').size().rename('Cant/jugadores').reset_index()
```

```python
df_joined
```

```python
#Opcional: Generar un gráfico que contenga, por año, el número de jugadores
import matplotlib.pyplot as plt
ax=df_joined.plot.bar(x='Joined',y='Cant/jugadores')
plt.show()
```
